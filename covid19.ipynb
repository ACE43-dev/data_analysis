### Day 5 — COVID-19, Our World in Data (Time-Series + Rolling Avg)
  - Filter to 2–3 countries, set date as index
  - 7-day rolling average of new cases (`.rolling(7).mean()`)
  - Plot rolling average, compare countries
  - Write takeaway: how waves differed in timing/peak
  

```python
df = pd.read_csv("DATASETSzip/full_grouped.csv")
```

```python
df.head()
```

```python
df.set_index("Date",inplace=True)
```

```python
df
```

```python
df[df["Country/Region"].isin(["India","China","Japan"])]
```

```python
df['cases_7day_avg'] = df['New cases'].rolling(7).mean()
```

```python
df[df["Country/Region"]=="China"]['cases_7day_avg'].head()
```

```python
df['cases_7day_avg'] = df.groupby('Country/Region')['New cases'].transform(lambda x: x.rolling(7).mean())
```

```python
df[df["Country/Region"]=="China"]['cases_7day_avg'].head()
```
